# Arithmetic Operations Tutorial

Welcome to the `concrete-fhe-toolkit` tutorials! This notebook exhaustively covers all bounded scalar arithmetic operations in `src/concrete_fhe_toolkit/arithmetic.py`.

In Fully Homomorphic Encryption (FHE), we cannot rely on native Python operators (`>`, `//`, etc.) because dynamic branching is impossible on encrypted data. We provide specialized, bounded cryptographic functions for every scenario.

<a id="comparing-values-compare"></a>
## Comparing Values (`compare`)
The `compare` function returns `1` when `x > y`, `0` when `x == y`, and `-1` when `x < y`. We test all three edge cases below.

```python
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compare, compile_compare

# 1. Cleartext Execution
assert compare(10, 5) == 1   # Greater
assert compare(5, 5) == 0    # Equal
assert compare(2, 8) == -1   # Less
print("Cleartext compare() passed all cases!")

# 2. FHE Compilation
# compile_compare handles bounds and builds the correct FHE circuit automatically
circuit = compile_compare(min_value=-15, max_value=15)

# 3. Encrypted Execution & Verification
assert circuit.encrypt_run_decrypt(10, 5) == 1
assert circuit.encrypt_run_decrypt(5, 5) == 0
assert circuit.encrypt_run_decrypt(2, 8) == -1
print("âœ… Encrypted compare() passed all edge cases!")
```

<a id="sign-of-a-value-sign"></a>
## Sign of a Value (`sign`)
The `sign` function returns the mathematical sign of a number (1, 0, or -1). Let's test it across positive, zero, and negative domains.

```python
from concrete_fhe_toolkit.arithmetic import sign, compile_sign

# 1. Cleartext Execution
assert sign(42) == 1
assert sign(0) == 0
assert sign(-10) == -1
print("Cleartext sign() passed all cases!")

# 2. FHE Compilation
circuit = compile_sign(min_value=-50, max_value=50)

# 3. Encrypted Execution & Verification
assert circuit.encrypt_run_decrypt(42) == 1
assert circuit.encrypt_run_decrypt(0) == 0
assert circuit.encrypt_run_decrypt(-10) == -1
print("âœ… Encrypted sign() passed all edge cases!")
```

<a id="floor-division-make_floor_divide"></a>
## Floor Division (`make_floor_divide`)
Division in FHE is notoriously difficult due to the lack of dynamic branching. `make_floor_divide` creates a function that performs exact floor division using multivariate table lookups. 

Crucially, we test what happens when **dividing by zero**, which is safely handled by our `zero_result` parameter!

```python
from concrete_fhe_toolkit.arithmetic import make_floor_divide, compile_floor_divide

# We want 0/0 or X/0 to yield 99 to prove zero-handling works securely!
cleartext_div = make_floor_divide(zero_result=99)

# 1. Cleartext Execution
assert cleartext_div(10, 3) == 3   # 10 // 3
assert cleartext_div(10, 5) == 2   # 10 // 5
assert cleartext_div(8, 0) == 99   # Division by zero handled!
print("Cleartext floor_divide passed all cases!")

# 2. FHE Compilation
circuit = compile_floor_divide(max_numerator=15, max_denominator=5, zero_result=99)

# 3. Encrypted Execution & Verification
assert circuit.encrypt_run_decrypt(10, 3) == 3
assert circuit.encrypt_run_decrypt(10, 5) == 2
assert circuit.encrypt_run_decrypt(8, 0) == 99
print("âœ… Encrypted floor_divide passed standard and zero-division edge cases!")
```

<a id="floor-divide-by-product-make_floor_divide_by_product"></a>
## Floor Divide by Product (`make_floor_divide_by_product`)
This is an advanced division where the denominator is the product of two numbers `numerator // (left * right)`. This is highly useful in financial calculations and complex scaling algorithms without consuming multiple FHE multiplication layers.

```python
from concrete_fhe_toolkit.arithmetic import make_floor_divide_by_product, compile_floor_divide_by_product

cleartext_div_prod = make_floor_divide_by_product(zero_result=44)

# 1. Cleartext Execution
assert cleartext_div_prod(12, 2, 2) == 3   # 20 // (2 * 3) = 20 // 6 = 3
assert cleartext_div_prod(12, 1, 0) == 44 # Division by zero handled!
print("Cleartext floor_divide_by_product passed all cases!")

# 2. FHE Compilation
circuit = compile_floor_divide_by_product(
    max_numerator=15, max_left=3, max_right=3, zero_result=44
)

# 3. Encrypted Execution & Verification
assert circuit.encrypt_run_decrypt(12, 2, 2) == 3
assert circuit.encrypt_run_decrypt(12, 1, 0) == 44
print("âœ… Encrypted floor_divide_by_product passed all edge cases!")
```
