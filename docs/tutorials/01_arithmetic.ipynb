# Arithmetic Operations in FHE
This notebook demonstrates completely isolated compilations of basic arithmetic operations.

## Basic Functions (`compare`, `sign`)

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compare

def test_cmp(x: int, y: int): return compare(x, y)
circ = fhe.Compiler(test_cmp, {"x": "encrypted", "y": "encrypted"}).compile([(-5, -5), (5, 5)])
assert circ.encrypt_run_decrypt(5, 3) == 1
assert circ.encrypt_run_decrypt(-5, 3) == -1
assert circ.encrypt_run_decrypt(0, 0) == 0
print('✅ Encrypted compare passed edge cases!')
```

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compile_compare

circ_comp = compile_compare()
assert circ_comp.encrypt_run_decrypt(5, 3) == 1
assert circ_comp.encrypt_run_decrypt(-5, 3) == -1
assert circ_comp.encrypt_run_decrypt(0, 0) == 0
print('✅ Encrypted compile_compare passed edge cases!')
```

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import sign

def test_sign(x: int): return sign(x)
circ = fhe.Compiler(test_sign, {"x": "encrypted"}).compile([(-42,), (42,)])
assert circ.encrypt_run_decrypt(42) == 1
assert circ.encrypt_run_decrypt(-7) == -1
assert circ.encrypt_run_decrypt(0) == 0
print('✅ Encrypted sign passed edge cases!')
```

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compile_sign

circ_sign = compile_sign()
assert circ_sign.encrypt_run_decrypt(42) == 1
assert circ_sign.encrypt_run_decrypt(-7) == -1
assert circ_sign.encrypt_run_decrypt(0) == 0
print('✅ Encrypted compile_sign passed edge cases!')
```

## Floor Division

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import make_floor_divide

fd = make_floor_divide(zero_result=99)
def test_fd(x: int, y: int): return fd(x, y)
circ = fhe.Compiler(test_fd, {"x": "encrypted", "y": "encrypted"}).compile([(-15, 0), (15, 5)])
assert circ.encrypt_run_decrypt(14, 3) == 4
assert circ.encrypt_run_decrypt(-14, 3) == -5
assert circ.encrypt_run_decrypt(0, 5) == 0
assert circ.encrypt_run_decrypt(10, 0) == 99
print('✅ Encrypted make_floor_divide passed!')
```

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compile_floor_divide

circ_fd = compile_floor_divide(max_numerator=15, max_denominator=10, zero_result=99)
assert circ_fd.encrypt_run_decrypt(14, 3) == 4
assert circ_fd.encrypt_run_decrypt(-14, 3) == -5 # Python floor division behavior
assert circ_fd.encrypt_run_decrypt(0, 5) == 0
assert circ_fd.encrypt_run_decrypt(10, 0) == 99 # Missing/zero_result
print('✅ Encrypted compile_floor_divide passed!')
```

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import make_floor_divide_by_product

fdp = make_floor_divide_by_product(zero_result=99)
def test_fdp(x: int, y: int, z: int): return fdp(x, y, z)
circ = fhe.Compiler(test_fdp, {"x": "encrypted", "y": "encrypted", "z": "encrypted"}).compile([(-15, -2, 0), (15, 2, 2)])
assert circ.encrypt_run_decrypt(12, 2, 2) == 3
assert circ.encrypt_run_decrypt(14, -2, 2) == -4
assert circ.encrypt_run_decrypt(0, 2, 2) == 0
assert circ.encrypt_run_decrypt(10, 0, 2) == 99
print('✅ Encrypted make_floor_divide_by_product passed!')
```

```
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compile_floor_divide_by_product

circ_fdp = compile_floor_divide_by_product(max_numerator=15, max_left=3, max_right=3, zero_result=99)
assert circ_fdp.encrypt_run_decrypt(12, 2, 2) == 3
assert circ_fdp.encrypt_run_decrypt(14, -2, 2) == -4
assert circ_fdp.encrypt_run_decrypt(0, 2, 2) == 0
assert circ_fdp.encrypt_run_decrypt(10, 0, 2) == 99
print('✅ Encrypted compile_floor_divide_by_product passed!')
```
