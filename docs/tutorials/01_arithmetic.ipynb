### Testing compare\nThis function securely compares two values. It returns 1 if x > y, 0 if x == y, and -1 if x < y. Positive, negative, and equality cases are tested.

```
import numpy as np
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compare

def test_compare(x, y):
    return compare(x, y)

compiler = fhe.Compiler(test_compare, {"x": "encrypted", "y": "encrypted"})
inputset = [(0, 0), (-2, -2), (3, -1), (-1, 3)]
circuit = compiler.compile(inputset)

test_cases = [((0, 0), 0), ((-2, -2), 0), ((3, -1), 1), ((-1, 3), -1)]
for (x, y), expected in test_cases:
    result = circuit.encrypt_run_decrypt(x, y)
    print(f"compare({x}, {y}) = {result} (Expected: {expected})")
    assert result == expected, f"Error: {result} != {expected}"
```

### Testing compile_compare\nThis is a directly compiled version of the compare function. min_value and max_value limits are provided.

```
import numpy as np
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import compile_compare

circuit = compile_compare(min_value=-3, max_value=3)
test_cases = [((0, 0), 0), ((-3, -3), 0), ((3, -1), 1), ((-1, 3), -1)]

for (x, y), expected in test_cases:
    result = circuit.encrypt_run_decrypt(x, y)
    print(f"compile_compare({x}, {y}) = {result} (Expected: {expected})")
    assert result == expected, f"Error: {result} != {expected}"
```

### Testing make_floor_divide\nPerforms floor division on encrypted values. The zero_result parameter is used to safely handle division by zero cases. Tests include division by 0.

```
import numpy as np
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import make_floor_divide

floor_div = make_floor_divide(zero_result=0)

def test_floor_divide(x, y):
    return floor_div(x, y)

compiler = fhe.Compiler(test_floor_divide, {"x": "encrypted", "y": "encrypted"})
inputset = [(4, 2), (-4, 2), (0, 3), (3, 0), (5, 2), (-5, 2)]
circuit = compiler.compile(inputset)

test_cases = [((4, 2), 2), ((-4, 2), -2), ((0, 3), 0), ((3, 0), 0), ((5, 2), 2), ((-5, 2), -3)]
for (x, y), expected in test_cases:
    result = circuit.encrypt_run_decrypt(x, y)
    print(f"floor_div({x}, {y}) = {result} (Expected: {expected})")
    assert result == expected, f"Error: {result} != {expected}"
```

### Testing make_floor_divide_by_product\nDivides the numerator by the product of two denominators. Limits are kept low to prevent overflow.

```
import numpy as np
from concrete import fhe
from concrete_fhe_toolkit.arithmetic import make_floor_divide_by_product

floor_div_prod = make_floor_divide_by_product(zero_result=0)

def test_floor_div_prod(num, d1, d2):
    return floor_div_prod(num, d1, d2)

compiler = fhe.Compiler(test_floor_div_prod, {"num": "encrypted", "d1": "encrypted", "d2": "encrypted"})
inputset = [(8, 2, 2), (-8, 2, 2), (0, 2, 2), (8, 0, 2), (5, 2, 1)]
circuit = compiler.compile(inputset)

test_cases = [((8, 2, 2), 2), ((-8, 2, 2), -2), ((0, 2, 2), 0), ((8, 0, 2), 0), ((5, 2, 1), 2)]
for (n, d1, d2), expected in test_cases:
    result = circuit.encrypt_run_decrypt(n, d1, d2)
    print(f"floor_div_prod({n}, {d1}, {d2}) = {result} (Expected: {expected})")
    assert result == expected, f"Error: {result} != {expected}"
```
