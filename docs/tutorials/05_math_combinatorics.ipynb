# Encrypted Math Tutorial: Combinatorics

This tutorial covers `src/concrete_fhe_toolkit/math/combinatorics.py`. Since factorial and combinations grow extremely fast, these functions use highly optimized lookup tables to bound the output values within FHE limits.

<a id="factorial-fibonacci-make_factorial-make_fibonacci"></a>
## Factorial & Fibonacci (`make_factorial`, `make_fibonacci`)

```python
from concrete import fhe
from concrete_fhe_toolkit.math.combinatorics import make_factorial, make_fibonacci

fact_fn = make_factorial(max_n=6)
fib_fn = make_fibonacci(max_n=10)

def test_sequences(n: int):
    return fact_fn(n), fib_fn(n)

assert test_sequences(5) == (120, 5) # 5! = 120, Fib(5) = 5
assert test_sequences(6) == (720, 8) # 6! = 720, Fib(6) = 8
print("Cleartext sequences passed!")

compiler = fhe.Compiler(test_sequences, {"n": "encrypted"})
inputset = [(0,), (3,), (6,)]
circuit = compiler.compile(inputset)

enc_res = circuit.encrypt_run_decrypt(5)
assert enc_res == (120, 5)
print("✅ Encrypted sequences passed!")
```

<a id="combinations-permutations-make_comb-make_perm"></a>
## Combinations & Permutations (`make_comb`, `make_perm`)

```python
from concrete_fhe_toolkit.math.combinatorics import make_comb, make_perm

comb_fn = make_comb(max_n=5, invalid_result=99)
perm_fn = make_perm(max_n=5, invalid_result=99)

def test_comb_perm(n: int, r: int):
    return comb_fn(n, r), perm_fn(n, r)

assert test_comb_perm(5, 2) == (10, 20)  # C(5,2)=10, P(5,2)=20
assert test_comb_perm(3, 5) == (99, 99) # r > n edge case!
print("Cleartext comb/perm passed!")

compiler = fhe.Compiler(test_comb_perm, {"n": "encrypted", "r": "encrypted"})
inputset = [(0, 0), (5, 2), (3, 5)]
circuit = compiler.compile(inputset)

enc_res1 = circuit.encrypt_run_decrypt(5, 2)
enc_res2 = circuit.encrypt_run_decrypt(3, 5)
assert enc_res1 == (10, 20)
assert enc_res2 == (99, 99)
print("✅ Encrypted comb/perm passed!")
```

<a id="power-and-powmod-make_power-make_powmod"></a>
## Power and PowMod (`make_power`, `make_powmod`)

```python
from concrete_fhe_toolkit.math.combinatorics import make_power, make_powmod

pow_fn = make_power(base=2, max_exponent=10)
powmod_fn = make_powmod(base=2, modulus=5, max_exponent=10)

def test_power(exp: int):
    return pow_fn(exp), powmod_fn(exp)

assert test_power(3) == (8, 3) # 2^3 = 8, 8 % 5 = 3
print("Cleartext power passed!")

compiler = fhe.Compiler(test_power, {"exp": "encrypted"})
inputset = [(0,), (3,), (10,)]
circuit = compiler.compile(inputset)

enc_res = circuit.encrypt_run_decrypt(3)
assert enc_res == (8, 3)
print("✅ Encrypted power passed!")
```
