# Encrypted Math Tutorial: Binary Division

This tutorial covers `src/concrete_fhe_toolkit/math/binary_division.py`. Division is notoriously difficult in FHE because we cannot use dynamic loops or branching. This module provides restoring division algorithms entirely at the bit level, enabling encrypted division and modulo.

<a id="unsigned-floor-division-make_unsigned_floor_divide"></a>
## Unsigned Floor Division (`make_unsigned_floor_divide`)

```python
from concrete import fhe
from concrete_fhe_toolkit.math.binary_division import make_unsigned_floor_divide

# Create division for 4-bit numbers (0 to 15)
div_fn = make_unsigned_floor_divide(numerator_width=4, denominator_width=4, zero_result=99)

def test_div(num: int, den: int):
    return div_fn(num, den)

assert test_div(10, 3) == 3
assert test_div(15, 4) == 3
assert test_div(10, 0) == 99  # Zero-division fallback
print("Cleartext unsigned division passed!")

compiler = fhe.Compiler(test_div, {"num": "encrypted", "den": "encrypted"})
inputset = [(0, 1), (10, 3), (10, 0)]
circuit = compiler.compile(inputset)

enc_res1 = circuit.encrypt_run_decrypt(10, 3)
enc_res2 = circuit.encrypt_run_decrypt(10, 0)
assert enc_res1 == 3
assert enc_res2 == 99
print("✅ Encrypted unsigned division passed!")
```

<a id="fixed-point-division-make_fixed_point_divide"></a>
## Fixed Point Division (`make_fixed_point_divide`)
Often you want a fractional result. This performs `(numerator << fractional_bits) / denominator`.

```python
from concrete_fhe_toolkit.math.binary_division import make_fixed_point_divide

fp_div_fn = make_fixed_point_divide(numerator_width=4, denominator_width=4, fractional_bits=2, zero_result=99)

def test_fp_div(num: int, den: int):
    return fp_div_fn(num, den)

# 5 / 2 = 2.5. With fractional_bits=2, (5 << 2) / 2 = 20 / 2 = 10.
assert test_fp_div(5, 2) == 10
print("Cleartext fixed point division passed!")

compiler = fhe.Compiler(test_fp_div, {"num": "encrypted", "den": "encrypted"})
inputset = [(0, 1), (5, 2)]
circuit = compiler.compile(inputset)

enc_res = circuit.encrypt_run_decrypt(5, 2)
assert enc_res == 10
print("✅ Encrypted fixed point division passed!")
```

<a id="unsigned-modulo-make_unsigned_mod"></a>
## Unsigned Modulo (`make_unsigned_mod`)

```python
from concrete_fhe_toolkit.math.binary_division import make_unsigned_mod

mod_fn = make_unsigned_mod(numerator_width=4, denominator_width=4, zero_result=99)

def test_mod(num: int, den: int):
    return mod_fn(num, den)

assert test_mod(10, 3) == 1
assert test_mod(15, 4) == 3
assert test_mod(10, 0) == 99
print("Cleartext modulo passed!")

compiler = fhe.Compiler(test_mod, {"num": "encrypted", "den": "encrypted"})
inputset = [(0, 1), (10, 3), (10, 0)]
circuit = compiler.compile(inputset)

enc_res = circuit.encrypt_run_decrypt(10, 3)
assert enc_res == 1
print("✅ Encrypted modulo passed!")
```
