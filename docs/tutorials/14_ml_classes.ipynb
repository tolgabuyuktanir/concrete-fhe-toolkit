# Machine Learning Module: Classes

This notebook tests the FHE compiler integration for the `classes` module.

### FHELogisticRegression

Tests the compilation and FHE prediction for `FHELogisticRegression`.

```python
from concrete_fhe_toolkit.ml.classes import FHELogisticRegression
from concrete import fhe
import numpy as np

model = FHELogisticRegression(weights=[3, 2], bias=-7)
dummy_inputset = [[0, 0], [0, 0]]
model.compile(dummy_inputset)

sample = [0, 0]
fhe_result = model.predict(sample)
clear_result = model.simulate(sample)

print("FHE Result:", fhe_result)
print("Clear Result:", clear_result)

if isinstance(fhe_result, (list, np.ndarray)) or hasattr(fhe_result, "__iter__"):
    np.testing.assert_array_equal(fhe_result, clear_result)
else:
    assert fhe_result == clear_result
```
