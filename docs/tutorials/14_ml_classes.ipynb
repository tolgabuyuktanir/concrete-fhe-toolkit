# Encrypted Machine Learning: Classes

This tutorial covers `src/concrete_fhe_toolkit/ml/classes.py`. This module provides the `FHEModel` base class and its various implementations. These classes handle compiling the FHE circuit and running predictions.

<a id="linear-and-logistic-regression"></a>
## Linear and Logistic Regression

```python
from concrete_fhe_toolkit.ml.classes import FHELogisticRegression, FHELinearRegression

log_model = FHELogisticRegression(weights=[3, 1], bias=-10)
lin_model = FHELinearRegression(weights=[10, -5], bias=2)

dummy_input = [[[0, 0]] * 16]
log_model.compile(dummy_input, batch_size=1)
lin_model.compile(dummy_input, batch_size=1)

assert log_model.predict([4, 1]) == 1
assert lin_model.predict([2, 1]) == 17
print("✅ Linear and Logistic Regression passed!")
```

<a id="trees-and-forests-decision-tree-random-forest-xgboost"></a>
## Trees and Forests (Decision Tree, Random Forest, XGBoost)

```python
from concrete_fhe_toolkit.ml.classes import FHEDecisionTree, FHERandomForest, FHEXGBoost

tree_def = {"feature": 0, "threshold": 5, "left": 1, "right": 0}

tree_model = FHEDecisionTree(tree=tree_def)
rf_model = FHERandomForest(trees=[tree_def, tree_def, tree_def])
xgb_model = FHEXGBoost(trees=[tree_def, tree_def]) # Normally XGBoost adds the trees

dummy_tree_input = [[[0]] * 16]
tree_model.compile(dummy_tree_input, batch_size=1)
rf_model.compile(dummy_tree_input, batch_size=1)
xgb_model.compile(dummy_tree_input, batch_size=1)

assert tree_model.predict([7]) == 1
assert rf_model.predict([3]) == 0
assert xgb_model.predict([7]) == 2 # 1 + 1
print("✅ Decision Tree, Random Forest and XGBoost passed!")
```

<a id="svm-and-knn"></a>
## SVM and KNN

```python
from concrete_fhe_toolkit.ml.classes import FHESVM, FHEKNN

svm_model = FHESVM(weights=[1, -1], bias=0)
svm_model.compile([[[0, 0]] * 16], batch_size=1)
assert svm_model.predict([5, 2]) == 1

knn_model = FHEKNN(X_train=[[0,0], [10,10]], y_train=[0, 1], k=1)
knn_model.compile([[[0, 0]] * 16], batch_size=1)
assert knn_model.predict([1, 1]) == 0
print("✅ SVM and KNN passed!")
```

<a id="multi-layer-perceptron-mlp-and-convolutional-neural-networks-cnn"></a>
## Multi-Layer Perceptron (MLP) and Convolutional Neural Networks (CNN)

```python
from concrete_fhe_toolkit.ml.classes import FHEMLP, FHECNN

# 1 hidden layer with 2 neurons, output layer with 1 neuron
w1 = [[1, 0], [0, 1]]
b1 = [0, 0]
w2 = [[1, 1]]
b2 = [0]

mlp_model = FHEMLP(mlp_layers=[(w1, b1), (w2, b2)])
mlp_model.compile([[[0, 0]] * 16], batch_size=1)
assert mlp_model.predict([2, 3])[0] == 5

# CNN (1 filter, 2x2 image, 2x2 kernel)
filter_1 = [[[1, 0], [0, 1]]]
bias = [0]
cnn_model = FHECNN(filters=filter_1, bias=bias)
cnn_model.compile([[[[0, 0], [0, 0]]] * 16], batch_size=1)
assert cnn_model.predict([[5, 0], [0, 3]]) == [[8]] # 5*1 + 3*1
print("✅ MLP and CNN passed!")
```

<a id="naive-bayes-pca-and-kmeans"></a>
## Naive Bayes, PCA and KMeans

```python
from concrete_fhe_toolkit.ml.classes import FHENaiveBayes, FHEPCA, FHEKMeans

# Naive Bayes: 2 classes, 1 binary feature
nb_model = FHENaiveBayes(log_prob_tables=[[[-1, -10]], [[-10, -1]]], priors=[0, 0])
nb_model.compile([[[0]] * 16], batch_size=1)
assert nb_model.predict([0]) == 0
assert nb_model.predict([1]) == 1

# PCA: 2 features -> 1 feature
pca_model = FHEPCA(means=[5, 5], components=[[1, 1]])
pca_model.compile([[[0, 0]] * 16], batch_size=1)
assert pca_model.predict([10, 10]) == [10] # (10-5)*1 + (10-5)*1

# KMeans: 2 centroids
kmeans_model = FHEKMeans(centroids=[[0, 0], [10, 10]], max_distance=300)
kmeans_model.compile([[[0, 0]] * 16], batch_size=1)
assert kmeans_model.predict([1, 1]) == 0
assert kmeans_model.predict([9, 9]) == 1

print("✅ Naive Bayes, PCA and KMeans passed!")
```
