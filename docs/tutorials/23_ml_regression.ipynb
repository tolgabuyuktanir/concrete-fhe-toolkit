# Machine Learning Module: Regression

This notebook tests the FHE compiler integration for the `regression` module.

### FHEXGBoostRegressor

Tests the compilation and FHE prediction for `FHEXGBoostRegressor`.

```python
from concrete_fhe_toolkit.ml.regression import FHEXGBoostRegressor
from concrete import fhe
import numpy as np

model = FHEXGBoostRegressor(trees=[{'feature': 0, 'threshold': 2, 'left_value': 1, 'right_value': 0}])
dummy_inputset = [[0, 0], [0, 0]]
model.compile(dummy_inputset)

sample = [0, 0]
fhe_result = model.predict(sample)
clear_result = model.simulate(sample)

print("FHE Result:", fhe_result)
print("Clear Result:", clear_result)

if isinstance(fhe_result, (list, np.ndarray)) or hasattr(fhe_result, "__iter__"):
    np.testing.assert_array_equal(fhe_result, clear_result)
else:
    assert fhe_result == clear_result
```
